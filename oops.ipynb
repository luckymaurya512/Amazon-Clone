***Encapsulation***

```python
class Account:
  def __init__(self, balance):
    self.__balance=balance

  def deposit(self, amount):
    if amount <= 0:
      print("Amount should be greater than 0")
    else:
      self.__balance += amount
      print("Amount deposited :", amount)

  def withdraw(self, amount):
    if amount <= self.__balance:
      self.__balance -= amount
      print("Amount withdrawn :", amount)
    else:
      print("You don't have enough balance")

  def get_balance(self):
    return "Account Balance : ", self.__balance

account = Account(1000)
account.get_balance()
```

```python
account.deposit(0)
print(account.get_balance())
```

```python
account.withdraw(10000)
```

```python
account.get_balance()
```

***Abstraction***

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    def eat(self):
        pass
class Cat(Animal):
    def sound(self):
        print("Cat : Mews")
class Dog(Animal):
    def sound(self):
        print("Dog : Barks")
    def eat(self):
        print("Eating...")
cat1 = Cat()
dog=Dog()
# cat1.eat()   # it won't run
```

```python
cat1.sound()
```

```python
dog.sound()
```

```python
dog.eat()
```

***Inheritance***

```python
class GrandParent:
  def wisdom(self):
    print("Be Polite")

class Father(GrandParent):
  def knowledge(self):
    print("Be healthy")

class Mother(GrandParent):
  def gyan(self):
    print("Work Hard")

class Child(Father, Mother):
  def fun(self):
    print("Be Happy")

child = Child()
```

```python
child.wisdom()
```

```python
child.knowledge()
```

```python
child.fun()
```

```python
child.gyan()
```

***Polymorphism***

```python
# Method Overriding

class Parent:
  def fun(self):
    print("Parent Class")

class Child(Parent):
  def fun(self):
    super().fun()   # if now used.. it will call child class directly
    print("Child Class")

child = Child()
```

```python
child.fun()
```

```python
# Method Overloading

class Calculator:
  def add(self, a, b, c=0):
    print(a+b+c)
  def sub(self, a, b):
    print(a-b)

cal = Calculator()
```

```python
cal.add(4,9)
```

```python
cal.sub(0,9)
```

```python
cal.add(344,55,23)
```

```python
# Dynamic Polymorphism / Duck Typing

class A:
  def fun(self):
    print("This is class A")

class B:
  def fun(self):
    print("This is class B")

class C:
  def fun(self):
    print("This is class C")

def fun2(obj):
  obj.fun()

fun2(A())
```

```python
fun2(B())
```

```python
fun2(C())
```

```python
# Operator overloading
class Book:
  def __init__(self, pages):
    self.pages = pages
  def __add__(self, other):
    return self.pages + other.pages

book1 = Book(100)
book2 = Book(200)
```

```python
book3=book1+book2
book3
```
